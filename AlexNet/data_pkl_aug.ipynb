```python
def create224_pkl(path,IMG_SIZE):
    dir=[]
    for i in tqdm(os.listdir(path)):
            dir.append(path+"\\"+i+"\\")
    print(dir)
    for i in dir:
        for e in range(1100):
            pat=i+str(e)
            class_num=e
            for dir_path in tqdm(os.listdir(pat)):
                #print(pat+"\\"+dir_path)
                img = cv2.imread(pat+"\\"+dir_path)
                data.append([img, class_num])
    random.shuffle(data)
    x = []
    y = []

    for features, label in tqdm(data):
        #print(label)
        x.append(features)
        y.append(label)
        
    y = np.array(y)
    y = y.astype(np.int32)
    x = np.array(x)
    x = x.astype(np.float32)

    x /= 255.0
    x = np.array(x).reshape(-1, 3, IMG_SIZE, IMG_SIZE)
    
    pickle_out = open("face224_x_features_partface.pkl","wb")
    pickle.dump(x,pickle_out,protocol=4)
    pickle_out.close()

    pickle_out = open("face224_y_features_partface.pkl","wb")
    pickle.dump(y,pickle_out,protocol=4)
    pickle_out.close()
```
